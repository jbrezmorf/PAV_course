## Řešení rovnic

```python
x=Symbol('x')
solve(x**2+5*x-6)
```

```python
# solve works for simple complex equations by default
x=Symbol('x')
eq = Eq(x**2-4*x,-5)
eq
```

```python
solve(eq)
```

```python
# Works for cubic roots as well.
solve(Eq(x**3,1))
```

```python
# Can restrict symbol to real numbers.
x=Symbol('x', real=True)
solve(Eq(x**3,1))
```

```python
# Not so good for more complex equations
x=Symbol('x', integer=True)
solve(Abs(x) < 2)
```

```python
# Still usefull for more complex abs equations
x=Symbol('x', real=True)
eq = Abs(x-3) > 2
eq
```

```python
# Still usefull for more complex abs equations
x=Symbol('x', real=True)
eq = sy.Abs(x-3) > 2
eq
```

```python
solve(eq)
```

```python
# solveset does not take symbol restrictions
solveset(eq)
```

```python
# Must set real domain explicitly
solveset(eq, domain=S.Reals)
```

## Rekurentní rovnice

```python
# Fibonacci sequence

n=Symbol('n')
f = Function('f')
Fib = f(n+2) - f(n+1) - f(n)
print('solving: 0 = ', Fib)
s = rsolve(Fib, f(n))
print('Explicit formula: ', s)
```

```python
# Time complexity of merge sort
# DOSN'T WORK
k=Symbol('k')
fe = Function('m')
merge = fe(k) - k - 2 * fe(k/2)
print('solving: 0 = ', merge)
s = rsolve(merge, f(k))
print('Explicit formula: ', s)
```

## Derivace

```python
# Differentiation
x=Symbol("x")
diff(sin(x), x )
```

```python
# Derivatives of vectors
import sympy.vector as vec

# 3d system coordinate vectors: i, j, k
S = vec.CoordSys3D('T')

r=Symbol("r")
phi=Symbol("phi")

# polar to cartesian
xy = r*cos(phi)*S.i + r*sin(phi)*S.j
display(xy)

# Jacobi matrix
dr = diff(xy, r)
dphi = diff(xy, phi)

[dr, dphi]
```

## Integrály

```python
# Integrate
x=Symbol("x")
k=Symbol("k")
n=Symbol("n", integer=True)
expr = integrate(cosh(x/k)*cos(x*pi*n), (x, 0,1))

# Result is Piecewise function:
display(expr)

# Get just second branch expression
formula = expr.args[1][0]
formula
```

```python
# Write formula as C code
print(ccode(formula))
print("\n=== pretty print ===\n")

# Pretty print
pprint(formula)

print("\n=== Sympy code ===\n")
# print Sympy code
print_python(formula)

print("\n=== Sympy code string ===\n")
# Make code in string. 
python(formula)
```

## Diferenciální rovnice

```python
# Differential equations
k=Symbol("k")
P2=Symbol("P2")
n=Symbol("n", integer=True)
f=Function('f')
deq = Eq(f(x).diff(x,x) + f(x), cos(n*pi*x))
display(deq)

# Get general solution
sol = dsolve(deq).rhs
display(sol)

# Solv boundary conditions
constants = solve([sol.diff(x).subs(x,0), sol.subs(x,1) - P2])
display(constants)

# substitute
sol.subs(constants[0])
```

## Vyhodnocování výrazů, generování kódu

```python

x, y = symbols('x y')
expr = 3*x**2 + log(x**2 + y**2 + 1)
expr
```

```python
from timeit import timeit

# Simplest method: using subs
def eval1():
    expr.subs({x: 17, y: 42}).evalf()
    
timeit(eval1, number=10000)
```

```python
import math 

# Using handmade function
def my_expr(x, y):
    x2 = x**2
    return 3* x2 + math.log(x2 + y**2 + 1)

def eval2():
    my_expr(17, 42)
    
timeit(eval2, number=10000)
```

```python
# Let SymPy to make the function from expression
g = lambdify([x, y], expr, modules=['math'])

def eval3():
    g(17, 42)
    
timeit(eval3, number=10000)    
```

Použití SymPy pro generování kódu funkcí ušetří mnoho manuální práce 
a vede na podobně rychlé vyhodnocení.
