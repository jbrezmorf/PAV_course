# Téma: Měření tíhového zrychlení z videa

**Vzdělávací cíl:** Porozumět odhadu fyzikální veličiny z měření a významu kalibrace a nejistoty. 

**Zadání a základ Pythonu:** Použijte krátký záznam padajícího nebo vyhozeného míčku se stojící kamerou a délkovým měřítkem v rovině pohybu. Pro základní řešení postačí CSV se sloupci `frame` a `y_px` jsou číslo snímku a vertikální souřadnice pixelu polohy předmětu  získané ručním odečtem alespoň pěti snímků během volného letu. 
Pro ruční odečet použijte např. editor GIMP, který zobrazuje polohu myši v pixelech a výšku středu míčku určete jako průměr výšky spodního a horního okraje míčku. Spolu s daty uveďte snímkovou frekvenci videa $f$ a měřítko $s$ v metrech na pixel. Alternativně můžete implementovat automatické sledování, viz volitelný bod níže.

Načtěte CSV tabulku, vypočtěte sloupce času $t_i=(n_i-n_0)/f$ $[s]$ a polohy $y_i=-s(p_i-p_0)$ $[m]$, kde $n_i = frame_i$ a pixelová souřadnice $p_i $ roste dolů. 
Pomocí `numpy.linalg.lstsq` proložte model $y(t)=a+bt+ct^2$ a určete $g=-2c$. 

Vykreslete snímky videa z tabulky spolu s predikovanou polohou $y(t)$ znázorněnou pomocí vodorovné čáry. 

Dále vykreslete graf proložené křivky a naměřených bodů a dále graf rezidua, t.j. rozdíl modelu a měření. 

Ověřte výpočet na syntetické trajektorii se známým $g$ a zopakujte odhad při samostatné změně $s$ a $f$ o 1 %; stručně vyhodnoťte citlivost.

- Přidejte možnost klikání na míček: Matplotlib: `button_press_event` poskytuje souřadnice `event.xdata`, `event.ydata`. Provedeš detekci míčku a její zobrazení,
- Přidejte detekci míčku:
  1. Převod z barevného prostoru RGB do prostoru HSV (Hue - odstín, Saturation - sytost barvy, Value - intenzita):
  2. Výpočet masky barevně blízkých pixelů, použije se numpy array $H$ pro zložku Hue a
     numpy array $S$ pro složku Saturation:
     $$
        \min\bigl(|H-H_0|,\;180-|H-H_0|\bigr)\leq \Delta H,
        \qquad S\geq S_{\min}.
     $$
     Referenční hodnota $H_0$ se určí z Hue pixelu, na který se kliklo.
     Minimální hodnota saturace $S_min$ může být pevná. Vzdálenost ve složce Hue je komplikovanější, protože odstín se chová jako úhel.
  3. Určení souvislé plochy a jejího těžiště, středu míčku, pomocí funkce OpenCV: `cv2.connectedComponentsWithStats`.
  4. Vykreslete detekovanou masku a těžiště do snímku, na který se klikalo.
  5. Detekujte ve všech snímcích, $H0$ interpolujte nebo určete z nejbližšího klikaného snímku.
