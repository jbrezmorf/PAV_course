# Téma: Pohodlné odpružení — pružiny, tlumení a rezonance

**Vzdělávací cíl:** Pochopit kmitání a kompromis mezi pohodlím posádky
a kontaktu kola s vozovkou. 

**Zadání a základ Pythonu:** Hmota $m$ představující část hmotnosti karoserie a nákladu připadající na jedno kolo je spojena s kolem pomocí pružiny s tuhostí $k$ a nezatíženou délkou $L$ a pomocí tlumiče s koeficientem $c$; „tuhé kolo“ o poloměru $r$ přesně sleduje předepsanou výšku vozovky $z(t)$.
Polohy $x(t)$ karoserie a $y(t)$ středu kola měříme jako absolutní výšky od společné referenční roviny, kladný směr je vzhůru. V základním modelu je poloha kola předepsána vztahem $y(t)=z(t)+r$. Pohybová rovnice karoserie je:

$$
m\ddot x=-c(\dot x-\dot y)-k(x-y-L)-mg.
$$

Na rovné vozovce $z=0$ je rovnovážná výška karoserie $x_\mathrm{rov}=r+L-mg/k$. Výchylku od této rovnováhy tedy představuje $v(t) = x-x_\mathrm{rov}$, nikoli samotné $x$. Pro $v$ pak dostáváme jednodušší rovnici:

$$
m\ddot v=-c(\dot v-\dot z)-k(v-z).
$$

Řešte rovnici numericky, např. pomocí `solve_ivp`,
všechny parametry a počáteční podmínky musí být snadno měnitelné pomocí vstupních proměnných nebo parametrů funkce. 
Porovnejte netlumený, podkriticky tlumený a nadkriticky tlumený pohyb; kritická hodnota je $c_\mathrm{krit}=2\sqrt{km}$. 

a. Pro rovnou vozovku $z(t)=0$, nenulovou počáteční výchylku $v(0)$ a nulovou počáteční rychlost.
b. Pro $z(t)=a\sin(\omega t)$ a pro několik frekvencí $\omega$. Pro porovnání vytvořte grafy výchylky a rychlosti v závislosti na čase. Pro jaké hodnoty parametrů $c$ a $k$ může $z(t)$ růst do nekonečna?

**Otevřené cíle:** 
- Ověřte periodu netlumeného pohybu $T=2\pi\sqrt{m/k}$. 
- Přidejte animaci.
- Jaké jsou realistické hodnoty parametrů?
- Přidejte hmotnost kola $m_2$ a druhou pružinu s tuhostí $K$ simulující pružnost pneumatiky. Hmotnost karoserie nyní označte $m_1=m$. Polohy $x,y,z$ mají stejný význam jako v základním modelu. Pro polohu karoserie platí stejná rovnice:
   $$
    m_1\ddot x= -c(\dot x-\dot y) - k(x-y-L) - m_1g,
   $$
ale $y(t)$ je nyní samostatná neznámá. Na kolo působí síly opačné k silám zavěšení působícím na karoserii, jeho tíha a normálová síla vozovky $N$:


   $$
    m_2\ddot y= c(\dot x-\dot y) + k(x-y-L) + N - m_2g.
   $$

   Nejprve předpokládejte trvalý kontakt a použijte $N=K(r+z-y)$, kde $K$ je "tuhost" pneumatiky. Experimentujte s parametry systému pro různé frekvence v $z(t)$.
- Model trvalého kontaktu je fyzikálně použitelný pouze dokud $N\geq0$.
  Zahrňte možnost ztráty kontaktu kola s vozovkou. Ve stejné soustavě nahraďte normálovou sílu vztahem

   $$
    N=K\max(r+z-y,0).
   $$

   Síla je tak při stlačení pneumatiky kladná a působí vzhůru, při oddělení je nulová; vozovka kolo nepřitahuje. 
- Simulujte přejezd hladké nerovnosti (hrbol nebo díra), porovnejte maximální zrychlení karoserie auta a dobu ustálení s předem zvolenou tolerancí (pohodlí pasažérů a nároky na pevnost konstrukce), hledejte vhodné nastavení podpružení a tlumení pro splnění limitu. Co by se stalo, bez zahrnutí modelu pneumatiky?
