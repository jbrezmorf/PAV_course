# Hod míčem 

**Vzdělávací cíl:** Porozumět předpokladům, za kterých úhel 45° maximalizuje dolet, a vlivu odporu vzduchu. 
    
**Zadání pro PAV:** Modelujte hod v rovině z polohy $(0,h)$ rychlostí $v_0$ pod úhlem $\alpha$ nad vodorovnou rovinou;
zem je $y=0$, osa $y$ míří vzhůru. Parametry zadávejte v SI. Implementujte analytické řešení bez odporu vzduchu a numerické řešení pohybových rovnic při odporu vzduchu:
\begin{align*}
\dot{\mathbf r}&=\mathbf v,\\
m\dot{\mathbf v}&=(0,-mg)-\tfrac12\rho C_D A|\mathbf v|\mathbf v
\end{align*}
v bezvětří, kde $m$ je hmotnost míče, $A$ jeho čelní plocha, $\rho$ hustota vzduchu a $C_D$ součinitel odporu. 
Pro numerické řešení použijte `scipy.integrate.solve_ivp`.
Ukončete let při průchodu $y=0$ směrem dolů. Funkce má vracet trajektorii, čas letu a dolet. Vykreslete několik trajektorií a dolet v závislosti na úhlu; nejlepší úhel stačí najít na jemné mřížce v $(0°,90°)$. Pro $h=0$ a nulový odpor ověřte dolet $R=v_0^2\sin(2\alpha)/g$ a maximum u 45°. 

**Otevřené cíle:** 
- Přidejte do rovnic vliv větru (rychlost míčku pro výpočet odporové síly je relativní vůči aktuální rychlosti větru.
  Pro počátek uvažujte konstantní rychlost větru.
- Doplňte grafy trajektorií např. jejich postupným vykreslováním a animací letícího míčku.
- Zkuste použít knihovnu Bokeh nebo Matplotlib pro vytvoření grafů s interaktivními prvky, např. posuvníky.
- Zkuste implementovat vodorovnou rychlost větru jako náhodný spojitý průběh $V(t)$, stejný v celém prostoru. Následující postup vytváří jeho aproximaci na konečném časovém intervalu:
  1. Zvolte střední rychlost $\bar V$, směrodatnou odchylku $\sigma$ a korelační čas $\tau>0$. Na pravidelné mřížce $t_j=j\Delta t$, $j=0,\ldots,N-1$, vygenerujte nezávislé reálné hodnoty $u_j$ z normálního rozdělení $N(0,1)$. Krok $\Delta t$ volte výrazně menší než $\tau$ a délku intervalu výrazně větší než $\tau$; mřížka musí pokrýt celý let.
  2. Převeďte $u_j$ Fourierovou transformací do frekvenční oblasti. Pro úhlovou frekvenci $\omega$ použijte spektrální hustotu
     $$
        S(\omega)=\frac{2\sigma^2\tau}{1+(\omega\tau)^2},
        \qquad
        C(r)=\frac{1}{2\pi}\int_{\mathbb R}S(\omega)e^{i\omega r}\,d\omega
             =\sigma^2e^{-|r|/\tau},
     $$
     kde $r$ je časový odstup a $C(r)$ požadovaná kovariance kolísání větru. Frekvence $\omega$ je v rad/s, $\tau$ a $r$ v sekundách; $\sigma$ je v m/s.
  3. Fourierovy koeficienty násobte odmocninou spektrální hustoty, protože ta popisuje druhé mocniny amplitud. Při výchozí normalizaci FFT v NumPy a zvoleném šumu $N(0,1)$ použijte filtr $\sqrt{S(\omega)/\Delta t}$. Lze pracovat s `rfft` a frekvencemi `2*np.pi*rfftfreq(N, d=dt)`; odpovídající inverzní transformace `irfft` s délkou $N$ zajistí reálný výsledek.
  4. Proveďte inverzní Fourierovu transformaci a přičtěte $\bar V$. Výsledkem je konečná periodická aproximace uvedené kovariance; průměr a směrodatná odchylka jedné realizace se mohou od $\bar V$ a $\sigma$ lišit. Mezi časovými uzly hodnoty lineárně interpolujte, čímž získáte spojitý průběh $V(t)$.
  5. Celý průběh vygenerujte před řešením pohybových rovnic; při vyhodnocování síly už používejte tutéž realizaci. Zobrazte průběh větru a porovnejte vliv různých hodnot $\sigma$ a $\tau$ na let míče.
- Oteřený výzkum: vliv rotace míče / střely.
